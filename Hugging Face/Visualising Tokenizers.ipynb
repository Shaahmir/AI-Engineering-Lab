## Getting Started with LLAMA

```python
from transformers import AutoTokenizer
```

```python
tokenizer = AutoTokenizer.from_pretrained(
    "meta-llama/Llama-3.1-8B"
)
```

```python
text = "Artificial Intelligence allows machines to mimic human intelligence through learning and reasoning."
```

### Encoding Tokens

```python
tokens = tokenizer.encode(text)
```

```python
print(tokens)
```

### Decoding Tokens

```python
text = tokenizer.decode(tokens)
```

```python
print(text)
```

### Batch Decoding

```python
batch_text = tokenizer.batch_decode(tokens)
```

```python
print(batch_text)
```

### Calculating Words and Tokens

```python
characters = len(text)
words = len(text.split(' '))
tokens_count = len(tokens)
print(f"There are {characters} characters, {words} words and {tokens_count} tokens.")
```

### Getting Vocabulary

```python
vocab = tokenizer.get_vocab()
added_vocab = tokenizer.get_added_vocab()
vocab_len = len(vocab)
```

```python
print(vocab)
```

```python
print(added_vocab)
```

```python
print(vocab_len)
```

## Getting Started with LLAMA Instruct Variant

```python
tokenizer_INSTRUCT = AutoTokenizer.from_pretrained(
    "meta-llama/Llama-3.1-8B-Instruct"
)
```

```python
messages = [
    {"role": "system", "content": "You are a helpful assistant"},
    {"role": "user", "content": "Tell a light-hearted joke that will resonate with a room full of data scientists."}
]
```

```python
prompt = tokenizer_INSTRUCT.apply_chat_template(
    messages,
    tokenize = False,
    add_generation_prompt = True
)
```

```python
print(prompt)
```

### QWEN Coder

```python
code = """
def greet(name):
  print("Hello ", name)
"""
```

```python
tokenizer_QWENCODER = AutoTokenizer.from_pretrained(QWENCODER)
```

```python
tokens_QWENCODER = tokenizer_QWENCODER.encode(code)
```

```python
print(tokens_QWENCODER)
```

```python
for token_QWENCODER in tokens_QWENCODER:
  print(f"{token_QWENCODER:<10}: {repr(tokenizer_QWENCODER.decode(token_QWENCODER))}")
```

```python
batch_text_QWENCODER = tokenizer_QWENCODER.batch_decode(tokens_QWENCODER)
```

```python
print(batch_text_QWENCODER)
```

```python
prompt_QWENCODER = tokenizer_QWENCODER.apply_chat_template(
    messages,
    tokenize = False,
    add_generation_prompt = True
)
```

```python
print(prompt_QWENCODER)
```
