## Importing the libraries

> **Note:** This notebook uses **Transformers v4.57.1** instead of the latest major release because **Transformers v5 removed several legacy pipeline tasks**, including `question-answering`, `summarization`, and `translation`. Since these pipelines are widely used for learning and provide a simple introduction to common NLP tasks, we'll use the latest stable v4 release throughout this section to ensure compatibility with the examples and a smoother learning experience.

```python
import torch
from transformers import pipeline
from diffusers import DiffusionPipeline
```

## Sentiment Analysis

```python
sentiment_analysis = pipeline(
    "sentiment-analysis",
    "nlptown/bert-base-multilingual-uncased-sentiment",
    device = "cuda"
)
```

```python
sentiment_analysis_out = sentiment_analysis(
    "This movie was amazing, and I can't wait for the next season."
)
```

```python
print(sentiment_analysis_out)
```

## Named Entity Recognition

```python
ner = pipeline(
    "ner",
    device = "cuda"
)
```

```python
ner_out = ner(
    "Sarah Johnson traveled from London to Tokyo on Monday to attend an AI conference."
)
```

```python
print(ner_out)
```

```python
for out in ner_out:
  print(f"{out["word"]:<10}: {out["entity"]}")
```

## Question Answering

```python
question = "What are Hugging Face pipelines?"

context = (
    "Hugging Face Pipelines provide a high-level API for performing inference "
    "with pretrained models across common tasks such as text generation, "
    "text classification, question answering, image classification, and more."
)
```

```python
question_answerer = pipeline(
    "question-answering",
    device = "cuda"
)
```

```python
question_answerer_out = question_answerer(
    question,
    context
)
```

```python
print(question_answerer_out)
```

## Text Summarizer

```python
test_summarizer = pipeline(
    "summarization",
    device = "cuda"
)
```

```python
text = """
Hugging Face Transformers is one of the most popular libraries for working with modern AI models.
It provides a simple and consistent interface for performing tasks such as text generation, summarization, translation, sentiment analysis and much more.
Instead of building models from scratch, developers and data scientists can use thousands of pretrained models available on the Hugging Face Hub.
The library integrates seamlessly with PyTorch and TensorFlow, making it easy to experiment, fine-tune, and deploy transformer models.
Its ease of use and extensive ecosystem have made it a standard tool for researchers, students, and AI engineers around the world.
"""
```

```python
test_summarizer_out = test_summarizer(
    text,
    min_length = 25,
    max_length = 50,
    do_sample = False
)
```

```python
print(test_summarizer_out)
```

## Translation

```python
translator = pipeline(
    "translation_en_to_fr",
    device = "cuda"
)
```

```python
translator_out = translator(
    "Hugging Face's Pipelines API is extremely easy to use, making it an excellent choice even for beginners."
)
```

```python
print(translator_out)
```

## Classification

```python
classifier = pipeline(
    "zero-shot-classification",
    device = "cuda"
)
```

```python
classifier_out = classifier(
    "Hugging Face's Transformers library is incredibly useful!",
    candidate_labels = ["technology", "sport", "politics"]
)
```

```python
print(classifier_out)
```
